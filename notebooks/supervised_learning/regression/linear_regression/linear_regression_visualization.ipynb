```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ml_from_scratch.supervised_learning.regression.linear_regression.models.linear_regression import LinearRegression
from pathlib import Path
```

```python
current = Path.cwd()
root = None
for directory in [current,*current.parents]:
    if (directory/"src").exists():
        root = directory
        break

```

```python
csv_file = root/"datasets"/"raw"/"house_prices_toy.csv"
df = pd.read_csv(csv_file)
X = df[["size_sqft"]].to_numpy()
y = df[["price_lakhs"]].to_numpy()
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X, y, color='blue', label='Data points')
plt.title('House Prices vs Size')
plt.xlabel('Size (sqft)')
plt.ylabel('Price (lakhs)') 
plt.show()
```

```python
model = LinearRegression(learning_rate=1e-7, n_iterations=1000)
model.fit(X,y)
y_pred = model.predict(X)
plt.figure(figsize=(8, 6))
plt.plot(X, y_pred, color='red', label='Regression Line')
plt.scatter(X, y, color='blue', label='Data points')
plt.title('House Prices vs Size')
plt.xlabel('Size (sqft)')
plt.ylabel('Price (lakhs)')
plt.legend()
plt.show()
```
